## 5. Outlier Detection using IQR Method
Identify outliers in numeric columns using the Interquartile Range (IQR) method to understand data anomalies.

```python
for col in numeric_col:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    Lower_bound = Q1 - 1.5 * IQR

    outliers = data[(data[col]< Lower_bound) | (data[col]>upper_bound)]
    print(f"column: {col}")
    print(f"Outliers : {outliers.shape[0]}")
```

```python
outliers_col = ['residential_assets_value', 'commercial_assets_value', 'bank_asset_value']

# Filter to include only columns that exist in the DataFrame
existing_outliers_col = [col for col in outliers_col if col in data.columns]

for col in existing_outliers_col:
    # Fill NaN values with 0 and add 1 to ensure all values are positive and finite for boxcox
    # This assumes that NaN in asset values can be interpreted as 0.
    # If a column is entirely NaN, this will result in a column of 1s, which boxcox can process.
    transformed_data = data[col].fillna(0) + 1

    # Check for any remaining non-positive values after transformation preparation
    if (transformed_data <= 0).any():
        print(f"Warning: Column '{col}' contains non-positive values even after filling NaNs and adding 1. Skipping Box-Cox.")
        continue

    # Check if the data is constant before applying boxcox
    if len(transformed_data.unique()) == 1:
        print(f"Warning: Column '{col}' contains constant values after transformation preparation. Skipping Box-Cox.")
        continue

    data[f"{col}_boxcox"], _ = stats.boxcox(transformed_data)
```

## 7. Apply Box-Cox Transformation
Apply Box-Cox transformation to normalize the distribution of asset values and handle outliers.

```python
for col in existing_outliers_col:
    # Only proceed if the boxcox transformed column exists
    boxcox_col_name = f"{col}_boxcox"
    if boxcox_col_name in data.columns:
        Q1 = data[boxcox_col_name].quantile(0.25)
        Q3 = data[boxcox_col_name].quantile(0.75)
        IQR = Q3 - Q1
        upper_bound = Q3 + 1.5 * IQR
        Lower_bound = Q1 - 1.5 * IQR

        outliers = data[(data[boxcox_col_name] < Lower_bound) | (data[boxcox_col_name] > upper_bound)]
        print(f"column: {col}")
        print(f"Outliers : {outliers.shape[0]}")
    else:
        print(f"Warning: Transformed column '{boxcox_col_name}' does not exist, skipping outlier check.")
```

## 13. Prepare Features and Target for Model Training
Prepare feature matrix (X) by removing redundant columns and encode the target variable (y) for model training.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder


le = LabelEncoder()

y = pd.Series(le.fit_transform(data['loan_status']))


print(y.head())
```

## 14. Split Data into Training and Testing Sets
Divide the data into training (80%) and testing (20%) sets for model development and evaluation.

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## 15. Train Logistic Regression Model
Build and train a Logistic Regression classifier to predict loan approval.

```python
Logistic_model = LogisticRegression(max_iter=1000)
model = Logistic_model.fit(X_train,y_train)
```

## 16. Make Predictions with Logistic Regression
Generate predictions on the test set using the trained Logistic Regression model.

```python
y_pred = model.predict(X_test)
```

## 17. Evaluate Logistic Regression on Test Set
Generate classification report to evaluate model performance on the test data.

```python
report = classification_report(y_test,y_pred)
print(f"{report}")
```

## 18. Evaluate Logistic Regression on Training Set
Generate classification report to evaluate model performance on the training data.

```python
y_pred_train = model.predict(X_train)
report = classification_report(y_train,y_pred_train)
print(f"{report}")
```

## 19. Train Decision Tree Model
Build and train a Decision Tree classifier with a maximum depth of 5 for interpretability.

```python
decision_tree_model = DecisionTreeClassifier(random_state=42,max_depth=5)
decision_tree_model.fit(X_train, y_train)
Y_Pred_DT = decision_tree_model.predict(X_test)
```

## 20. Evaluate Decision Tree on Test Set
Generate classification report to evaluate Decision Tree model performance on test data.

```python
report = classification_report(y_test,Y_Pred_DT)
print(f"{report}")
```

## 21. Evaluate Decision Tree on Training Set
Generate classification report to evaluate Decision Tree model performance on training data.

```python
Y_Pred_DT = decision_tree_model.predict(X_train)
report = classification_report(y_train,Y_Pred_DT)
print(f"Report on Training {report}")
```

## 22. Cross-Validation Analysis
Perform 5-fold cross-validation on the Decision Tree model to assess model stability and reliability across different data splits.

```python
cross_validation = cross_val_score(decision_tree_model,x,y,cv=5, scoring='accuracy')
print(f"Average Accuracy is: {cross_validation.mean()}")
```

## 23. Hyperparameter Tuning for Decision Tree Model

We will use `GridSearchCV` to find the optimal hyperparameters for our `DecisionTreeClassifier`. This involves defining a range of parameters to test and then evaluating each combination using cross-validation to identify the best performing set.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

# Define the parameter grid to search
param_grid = {
    'max_depth': [3, 5, 7, 10, None], # Max depth of the tree
    'min_samples_split': [2, 5, 10], # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4], # Minimum number of samples required to be at a leaf node
    'criterion': ['gini', 'entropy'] # Function to measure the quality of a split
}

# Initialize a DecisionTreeClassifier
dt_classifier = DecisionTreeClassifier(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=dt_classifier, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1, verbose=1)

# Fit GridSearchCV to the training data
grid_search.fit(X_train, y_train)

# Print the best parameters and best score
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best cross-validation accuracy: {grid_search.best_score_:.4f}")
```

## 24. Evaluate Tuned Decision Tree Model

Now, let's evaluate the performance of the Decision Tree model with the best hyperparameters found by `GridSearchCV` on the test set.

```python
# Get the best model from GridSearchCV
best_dt_model = grid_search.best_estimator_

# Make predictions on the test set
y_pred_tuned_dt = best_dt_model.predict(X_test)

# Evaluate the tuned model
report_tuned = classification_report(y_test, y_pred_tuned_dt)
print(f"Classification Report for Tuned Decision Tree Model on Test Set:\n{report_tuned}")
```
